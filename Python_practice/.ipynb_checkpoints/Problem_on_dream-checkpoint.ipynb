```python
Array = [[12345698321465],
         [1010101],
         [1212121212],
         [326532653265],
         [2525252525],
         [11111111]]

# Ans = [0,0,1,0,1,0]
```

```python
# if the pait of inc is not in the list then false

def pair(lst):
    ans = []
    for i in lst:
        num_str = str(i[0]) 
        if len(num_str) % 2 != 0:
            ans.append(0)
            continue
        first, second = num_str[0], num_str[1]
        valid = True
        for j in range(len(num_str)):
            if (j % 2 == 0 and num_str[j] != first) or (j % 2 == 1 and num_str[j] != second):   
            # indexing bruhh so even is 1st positon (ie. 0th index is even) 
                valid = False
                break
        ans.append(1 if valid else 0)
    return ans

print(pair(Array))
                
            
            
            
```
