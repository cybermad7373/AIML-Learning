```python
# import lib
```

```python
import pandas as pd
```

```python
# import csv
```

```python
df = pd.read_csv('./data/CricketTestMatchData.csv')
df
```

```python
df['Span'] = df['Span'].astype(str)
df['Span'] = df['Span'].str.replace(r'(\d{4})(\d{4})', r'\1-\2', regex=True)
```

```python
# rename multiple column
```

```python
df = df.rename(columns = {'Player':'Player_Name', 'Span':'Playing_span',	'Mat':'matches_played', 'Inns':'Innings_batted', 'NO':'Not_outs', 
                          'Runs':'Runs_scored','HS' : 'Highest_inns_score',	'Ave':'Batting_average', 'BF':'Balls_faced', 'SR':'Batting_strike_rate', 
                          '100' : 'Hundreds_scored', '50':'Fifties_scored',	'0':'Ducks_scored',	'4s':'Boundary_fours',	'6s':'Boundary_sixes'})
```

```python
df.head()
```

```python
# check null value
```

```python
df.isna().any()
```

```python
df.isna().sum()
```

```python
df[df['Balls_faced'].isna()==1]
```

```python
df['Balls_faced'] = df['Balls_faced'].fillna(0) 
```

```python
df['Ducks_scored'] = df['Ducks_scored'].fillna(0) 
```

```python
#drop duplicates
```

```python
df.duplicated().any()
```

```python
df.duplicated().sum()
```

```python
df[df['Player_Name'].duplicated() == 1]
```

```python
df[df['Player_Name'].isin(['H Sutcliffe (ENG)','SM Gavaskar (IND)','KC Bland (SA)','AB de Villiers (SA)'])]
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().any()
```

```python
# spliting Playing_span to start and end date + carer length
```

```python
df['Playing_span'].str.split('-')
```

```python
df['Starting_year'] = df['Playing_span'].str.split('-').str[0]
df['Final_year'] = df['Playing_span'].str.split('-').str[1]
```

```python
df.head()
```

```python
#droping Playing_span column
```

```python
df.drop('Playing_span', axis = 1,inplace = True)
# df.drop(['Playing_span'], axis = 1,inplace = True)     # to drop multiple column use [] 

# cann use inplace = true / df = df.drop()   both will save the change 
```

```python
df.head()
```

```python
# Extract the country code and save it in a new column
df['Country'] = df['Player_Name'].str.extract(r'\((\w+)\)')

# Remove the country code from the original column
df['Player_Name'] = df['Player_Name'].str.replace(r'\s*\(\w+\)', '', regex=True)
```

```python
df.sample(3)
```

```python
# # other method
# # country
# df['Player_Name'].str.split('(')
# df['Country'] = df['Player_Name'].str.split('(').str[1]
# df['Country'] = df['Country'].str.split(')').str[0]

# # player name
# df['Player_Name'] = df['Player_Name'].str.split('(').str[0]
```

```python
# change data types
```

```python
df.dtypes
```

```python
df['Highest_inns_score'] = df['Highest_inns_score'].str.split('*').str[0]
df['Highest_inns_score'] = df['Highest_inns_score'].astype(int)
```

```python
# df['Balls_faced'] = df['Balls_faced'].astype('int')
# df['Boundary_fours'] = df['Boundary_fours'].astype('int')
# df['Boundary_sixes'] = df['Boundary_sixes'].astype('int')
```

```python
df['Boundary_fours'] = df['Boundary_fours'].str.split('+').str[0]
df['Boundary_sixes'] = df['Boundary_sixes'].str.split('+').str[0]
```

```python
df = df.astype({'Boundary_fours':'int', 'Boundary_sixes' : 'int', 'Starting_year':'int', 'Final_year':'int'})
```

```python
df.dtypes
```

```python
df['Career_length'] = df['Final_year'] - df['Starting_year']
```

```python
df.head()
```

```python
# Q1 : Avg career length
```

```python
df['Career_length'].mean()
```

```python
# Q2 : avg strike rate for player with 10+ years exp
```

```python
df[df['Career_length'] > 10 ]['Batting_strike_rate'].mean()
```

```python
# q3 : no of cricketers plyed before 1960
```

```python
df[df['Starting_year'] < 1960]['Player_Name'].count()
```

```python
# q4 : max high inn score by a coutry
```

```python
df.groupby('Country')['Highest_inns_score'].max()
```

```python
# q5 : hundreds, fifties, ducks AVG by country
```

```python
df.groupby('Country')[['Hundreds_scored', 'Fifties_scored', 'Ducks_scored' ]].mean()
```
