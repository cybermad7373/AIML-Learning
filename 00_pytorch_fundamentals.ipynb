### Reshaping, viewing and stacking, Squeezing, unsqueezing and permuting

image.png

```python
 import torch
 x = torch.arange(1.,11.)
 x , x.shape
```

```python
#adding extra dimension
x_reshape = x.reshape(5,2)
x_reshape, x_reshape.shape
```

```python
#changing th view
z = x.view(1,10)     # changing z will affect x value as well, cause it shares same memory
z , z.shape
```

```python
z[:, 0] = 354
z , x
```

```python
# Stack tensor on top

# x_stacked = torch.stack([x,x,x,x])    # Vertical stacking  `dim = 0 by default`
x_stacked = torch.stack([x,x,x,x], dim = 1)   #  horizontal stacking
x_stacked
```

```python
# Squeeze - removes all single diemension from the target tensor

z.shape , z
```

```python
z.squeeze() , z.squeeze().shape
```

```python
print(z.unsqueeze(dim = 0).shape)
print(z.unsqueeze(dim = 1).shape)
print(z.unsqueeze(dim = 2).shape)
```

```python
# permute - rearrange the dimension of the target in specified order
x_original = torch.rand(size = (244,244,3 ))  # [Hight, Width ,Color]

#permute original tensor to rearrange the axis order
x_permuted = x_original.permute(2,0,1)   # [ color , Height, Width]
x_original.shape , x_permuted.shape
```

```python
x_original[0,0,0] = 777
x_permuted[0,0,0], x_original[0,0,0]      # same , it shares same memory thoo
```

## Reproduciability (trying to take the random  out of random)

```python
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)
random_tensor_A, random_tensor_B
```

```python
random_tensor_A == random_tensor_B
```

```python
# random but reproduciable tensors
RANDOM_SEED = 10
torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C, random_tensor_D)

random_tensor_C == random_tensor_D
```
